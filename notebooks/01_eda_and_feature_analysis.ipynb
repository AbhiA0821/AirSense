# Milestone 1: Exploratory Data Analysis & Feature Importance Analysis
### AI-Powered Environmental Intelligence System

This notebook conducts a comprehensive Exploratory Data Analysis (EDA) on the Delhi Air Quality and Weather dataset (`merged_delhi_air_quality_weather.csv`).

#### Key Objectives:
1. **Dataset Overview & Missing Values Inspection**: Assess data completeness across pollutants and weather metrics.
2. **Target Variable Analysis (`AQI`)**: Distribution, categories (`Good`, `Satisfactory`, `Moderate`, `Poor`, `Very Poor`, `Severe`).
3. **Pollutants & Weather Correlation**: Inter-pollutant dependencies and environmental factor impacts.
4. **Temporal Patterns**: Diurnal (hourly) cycles and seasonal (monthly) trends.
5. **Spatial Analysis**: Air quality hotspot comparisons across 27 monitoring stations in Delhi.
6. **Feature Importance Ranking**: Random Forest and Mutual Information rankings to identify top predictor features for future AQI model development.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestRegressor
from sklearn.feature_selection import mutual_info_regression

# Set styling
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.size'] = 10

RAW_DATA_PATH = '../dataset/dataset.csv'
print('Loading raw dataset...')
```

```python
df = pd.read_csv(RAW_DATA_PATH)
print(f"Shape: {df.shape}")
df.head()
```

## 1. Missing Values Overview

```python
missing_summary = pd.DataFrame({
    'Dtype': df.dtypes,
    'Missing_Count': df.isnull().sum(),
    'Missing_Percent': (df.isnull().sum() / len(df)) * 100
}).sort_values('Missing_Percent', ascending=False)
missing_summary
```

## 2. Target Variable Analysis (`AQI`)

```python
df_valid = df.dropna(subset=['AQI']).copy()
df_valid['Datetime'] = pd.to_datetime(df_valid['Datetime'], format='%d-%m-%Y %H:%M', errors='coerce')
df_valid['Year'] = df_valid['Datetime'].dt.year
df_valid['Month'] = df_valid['Datetime'].dt.month
df_valid['Day'] = df_valid['Datetime'].dt.day
df_valid['Hour'] = df_valid['Datetime'].dt.hour
df_valid['DayOfWeek'] = df_valid['Datetime'].dt.dayofweek
df_valid['IsWeekend'] = df_valid['DayOfWeek'].isin([5, 6]).astype(int)

print("AQI Summary Statistics:")
print(df_valid['AQI'].describe())
```

## 3. Feature Importance Analysis for AQI Prediction

```python
sample_df = df_valid.dropna(subset=['PM2.5', 'PM10', 'NO2', 'CO', 'O3', 'SO2', 'NH3', 'temp (°C)', 'Relative Humidity (%)', 'Average Wind Speed (km/h)']).sample(min(50000, len(df_valid)), random_state=42)
candidate_features = ['PM2.5', 'PM10', 'NO', 'NO2', 'NOx', 'NH3', 'CO', 'SO2', 'O3', 'Benzene', 'Toluene',
                      'temp (°C)', 'Dew Point (°C)', 'Relative Humidity (%)', 'Average Wind Speed (km/h)',
                      'Sea-Level Air Pressure (hPa)', 'Hour', 'Month', 'DayOfWeek', 'IsWeekend']

X_sample = sample_df[candidate_features]
y_sample = sample_df['AQI']

rf = RandomForestRegressor(n_estimators=50, random_state=42, n_jobs=-1)
rf.fit(X_sample, y_sample)
mi_scores = mutual_info_regression(X_sample, y_sample, random_state=42)

feature_imp_df = pd.DataFrame({
    'Feature': candidate_features,
    'RandomForest_Importance': rf.feature_importances_,
    'MutualInformation_Score': mi_scores
}).sort_values('RandomForest_Importance', ascending=False)

feature_imp_df
```
